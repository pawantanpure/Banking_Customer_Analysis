```python
#Group customers by estimated income

bins = [0, 100000, 300000, float("inf")]
labels = ["Low", "Medium", "High"]

df["Income Band"] = pd.cut(
    df["Estimated Income"],
    bins=bins,
    labels=labels,
    right=False
)

print(df["Income Band"].value_counts())
```

```python
#Count customers in each income group

df["Income Band"].value_counts().plot(kind="bar")
plt.title("Customers by Income Group")
plt.xlabel("Income Group")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

```python
#Count values in category columns

category_columns = [
    "BRId", "GenderId", "IAId", "Amount of Credit Cards",
    "Nationality", "Occupation", "Fee Structure",
    "Loyalty Classification", "Properties Owned",
    "Risk Weighting", "Income Band"
]

for col in category_columns:
    if col in df.columns:
        print("\\n", col)
        print(df[col].value_counts(dropna=False))
```

```python
#Univariate analysis
#A bar chart for each category column. This helps us see which values are common.

for col in category_columns:
    if col in df.columns:
        df[col].value_counts().plot(kind="bar", figsize=(7, 4))
        plt.title("Count of " + col)
        plt.xlabel(col)
        plt.ylabel("Count")
        plt.tight_layout()
        plt.show()
```

```python
#Compare categories by nationality

for col in category_columns:
    if col in df.columns and col != "Nationality" and "Nationality" in df.columns:
        pd.crosstab(df[col], df["Nationality"]).plot(kind="bar", figsize=(8, 4))
        plt.title(col + " by Nationality")
        plt.xlabel(col)
        plt.ylabel("Number of Customers")
        plt.tight_layout()
        plt.show()
```

```python
#Numerical analysis

numerical_columns = [
    "Estimated Income", "Superannuation Savings", "Credit Card Balance",
    "Bank Loans", "Bank Deposits", "Checking Accounts",
    "Saving Accounts", "Foreign Currency Account", "Business Lending"
]

for col in numerical_columns:
    if col in df.columns:
        df[col].plot(kind="hist", bins=20, figsize=(7, 4))
        plt.title("Distribution of " + col)
        plt.xlabel(col)
        plt.ylabel("Number of Customers")
        plt.tight_layout()
        plt.show()
```

```python
#Correlation heatmap

available_numeric_columns = [col for col in numerical_columns if col in df.columns]
correlation = df[available_numeric_columns].corr()

plt.figure(figsize=(10, 8))
plt.imshow(correlation, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="Correlation")
plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=90)
plt.yticks(range(len(correlation.columns)), correlation.columns)
plt.title("Correlation Between Numeric Columns")
plt.tight_layout()
plt.show()
```
