```python
#import library
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the bank data

df = pd.read_csv(r"C:\Users\tanpu\Downloads\Banking.csv")


print(df.head())
```

```python
# Number of rows and columns
print('Rows and columns:', df.shape)

# Column names and data types
print(df.info())
```

```python
# Show estimated income
print(df['Estimated Income'].head())
```

```python
# Create income groups
# Low: up to 100,000; Mid: 100,001 to 300,000; High: above 300,000
def income_group(income):
    if income <= 100000:
        return 'Low'
    elif income <= 300000:
        return 'Mid'
    else:
        return 'High'

df['Income Band'] = df['Estimated Income'].apply(income_group)
print(df[['Estimated Income', 'Income Band']].head())
```

```python
#Count categories

print(df['Nationality'].value_counts())
print(df['Occupation'].value_counts())
print(df['Fee Structure'].value_counts())
print(df['Loyalty Classification'].value_counts())
print(df['Income Band'].value_counts())
```

```python
# Summary of numerical columns
print(df.describe())
```

```python
# Check missing values in each column
print(df.isnull().sum())
```

```python
# Convert Joined Bank to a date
# This assumes dates are written like 31-12-2020.
df['Joined Bank'] = pd.to_datetime(df['Joined Bank'], format='%d-%m-%Y', errors='coerce')
print(df['Joined Bank'].head())
```

```python
# Histogram: show the age distribution
df['Age'].plot(kind='hist', title='Customer Ages', bins=10)
plt.xlabel('Age')
plt.show()
```

```python
# Histogram: show estimated income distribution
df['Estimated Income'].plot(kind='hist', title='Estimated Income', bins=10)
plt.xlabel('Estimated Income')
plt.show()
```

```python
# Correlation between numerical columns
# Values close to 1 mean a strong positive relationship.
# Values close to -1 mean a strong negative relationship.
# Values close to 0 mean a weak relationship.
columns = ['Age', 'Estimated Income', 'Superannuation Savings', 'Credit Card Balance',
           'Bank Loans', 'Bank Deposits', 'Checking Accounts', 'Saving Accounts',
           'Foreign Currency Account', 'Business Lending', 'Properties Owned']
print(df[columns].corr().round(2))
```

```python
# Compare bank deposits and savings accounts
df.plot(kind='scatter', x='Bank Deposits', y='Saving Accounts', title='Bank Deposits vs Saving Accounts', alpha=0.4)
plt.show()
```

```python
# Compare age and superannuation savings
df.plot(kind='scatter', x='Age', y='Superannuation Savings', title='Age vs Superannuation Savings', alpha=0.4)
plt.show()
```
